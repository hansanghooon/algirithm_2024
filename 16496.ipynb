```python
# 정답

#1,000,000,000 이하라서 10번째까지 해줘야됨. 글에 9자리 -> 10자리로 바꿔서 보면 되는대 사실 몇자리인지는 논리적으로 크게 중요한 사실은 아님


import sys
import heapq
import math

input=sys.stdin.readline


# 최대 1,000,000,000 이하, 즉 모든 숫자는 1자리에서  9자리 숫자
#  숫자를 선택해 이어붙여 가며 가장 큰 숫자를 만드는 방식으로 문제를 해결
#  i번째 원소가 i+1 자리의 숫자 (0번째 인덱스는 1,2,3 등의 1자리수, 1번째 인덱스는 12,42 등 2자리수)의 숫자들이 담긴 heapq를 원소로 하는 리스트를 만듬
# 숫자를 담을 때 모든 숫자를 9자리 숫자로 만들것임(비교연산을 통해 문제를 해결하기 위함)-> 숫자의 자릿수가 9보다 작을 경우 자기 자신을 이어붙이고 9자리로 자름
# ex) 123 -> 123123123 , 4252-> 425242524252 (9자리로 자름)-> 425242524
# 자기 자신을 붙이는 이유는 일종의 기댓값의 개념, 해당 숫자가 선택되어 붙여질 경우,뒤에 붙여지는 수의 최대값은 해당 숫자보다 클수 없음,

# 해당 방식의 선택 이유
# 만약 맨 앞의 숫자로 87 이라는 숫자가 선택이 되는 순간을 생각해보자. 어느 상황에서 87이 선택될 수있을까?
# 87이 맨 앞에 숫자로 선택되기 위해서는 모든 숫자들의 앞에 두 자리수는 87 보다 작거나 같아야함 
# (만약 9~, 88~ ,89~, 로 시작하는 숫자들이 있으면 해당 숫자가 선택될것임)
# 87이 맨앞에 온후, 이후 숫자를 만들어가는 상황을 생각해보자 가장 최대가 될 때는 878787~ 의 형태가 된다.(선택한 숫자 87이 반복되는 형식)
# 만약 879 같이 3번째 자리의 숫자에 8보다 큰값이 존재하는 숫자가 있을 경우, 87은 선택될 수 없다 (87987('879'+'87')) > 87879 ('87'+'879') )
# 또한 같은 이유로 8788이나 8789 같이 3번째 자리 숫자까지 같아도 4번째 자리의 숫자가 87의 두번째 자리의 숫자인 7보다 클경우 87은 선택될수 없음

# 위의 논지를 n자리 숫자로 확장할 경우 
# 숫자 x 의 가장 높은 자릿수를 1로 생각하고 i번째 자릿수자를 x[i] 라고하고(x=921 일 경우 x[1]=9,x[2]=2) ceil(log10(x))=n 이라 하자.
# x가 가장 왼쪽으로 선택되어 가장 큰 수 X 를 만든다고 할 경우
# X[i]<=x[i%n] 의 형태가 되어야 한다.
# 위의 식을 보면 X가 최대가 될 경우 숫자 x를 계속 이어붙이는 경우임을 알 수 있다.



def make_bignum(num_list):
    # 0 담을곳이랑 다른거랑 비교 
    # minheap으로 구현되어 있기 때문에 최대값만 꺼내어 비교하기 위해 저장시 -1 을 곱해 저장할것임
    # nan 방지 위해서  1 넣어놓음
    
    if max(num_list)==0:
        return 0
    
    # heapq.heapify 는 return을 하지 않음
    heapq_list=[[1] for __ in range(10)]
    for element in heapq_list:
        heapq.heapify(element)

    for number in num_list:
        number=int(number)
        if number !=0:
            log_val=math.floor(math.log10(number))
            number=number*-1
            heapq.heappush( heapq_list[log_val],number)
        else:
            heapq.heappush( heapq_list[0],number)


    # 모든값을은 -1 이곱해져 저장되기 때문에 숫자를 입력받아 들어가는 값은 0보다 작거나 같음
    # check_out 이 1 이 되면, 즉 모든 heap에 nan방지를 위한 값 1만 남아서 모든값의 최소값이 1이 되면 종료
    
    
    #모든 heapq 에있는 숫자들 가져와서 숫자를 변형시켜준 뒤 비교시작
    choose=[heapq_list[heapq_idx][0] for heapq_idx in range(len(heapq_list)) ]

    for num_idx in range(len(choose)):

        number=choose[num_idx]
        # 값이 안담겨서 1 이 존재할경우 continue
        if number == 1:
            number=number*-1
            choose[num_idx]=number
            continue

        # 값이 들어있을 경우 
        # 음수 양수로 복구 
        number=number*-1
        number=str(number)
        # 길이 10 편의상 한 구현, 다른  방식으로 만들고 잘라도 될수도?
        number=number*10
        number=number[:10]
        number=int(number)
        choose[num_idx]=number


    #이제 숫자 하나씩 빼면서 빼는 숫자만 벌충해줄 것임

    res=''
    for a in range(len(num_list)):
        max_idx=choose.index(max(choose))
        add_num=heapq.heappop(heapq_list[max_idx])

        #choose 업데이트

        number=heapq_list[max_idx][0]
        if number == 1:
            number=number*-1
            choose[max_idx]=number
    
        else:
            # 값이 들어있을 경우 
            # 음수 양수로 복구 
            number=number*-1
            number=str(number)
            # 최소 길이 10 편의상 한 구현, 다른  방식으로 만들고 잘라도 될수도?
            number=number*10
            number=number[:10]
            number=int(number)
            choose[max_idx]=number
        add_num=add_num*-1
        add_str=str(add_num)
        res=res+add_str

    return res





a=input()


num_list=list(map(int,input().split()))

ans=make_bignum(num_list)

print(int(ans))
```
